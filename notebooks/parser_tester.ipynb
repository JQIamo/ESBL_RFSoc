```python
import socket

# Client Code
host="192.168.2.99"
port=6750
client = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
try:
    client.connect((host, port))
    print(f"[INFO] Connected to server {host}:{port}")

    message = "FC 1000000"
    # message = "FM 10000"
    # message = "MD 82"
    # message = "SA 0.99"
    # message = "CA 1.01"
    # message = "PM 2.0" TODO: ERROR
    # message = "SP 0.98"
    # message = "CLDMA"
    # message = "STDMA"
    # message = "ASC 1000000, 2000000, 100000;"
    # message = "AM 1.1"
    # message = "CLLOG"
    # message = "RDLOG"
    # message = "END"
    # message = "Q?"
    message = "SHUTDOWN"
    # message = "HELP"

    client.sendall(f"{message};".encode("ascii"))
    response = client.recv(4096).decode("ascii")
    print(f"[RESPONSE] {response}")

except Exception as e:
    print(f"[ERROR] {e}")
finally:
    client.close()
    print("[INFO] Client disconnected")
```
