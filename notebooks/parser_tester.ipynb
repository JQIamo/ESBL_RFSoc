```python
import socket

# Client Code
host="192.168.2.99"
port=6750
client = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
try:
    client.connect((host, port))
    print(f"[INFO] Connected to server {host}:{port}")

    message = "FC 1000000000"
    # message = "FM 625000"
    # message = "MD 82"
    # message = "SA 0.99"
    # message = "CA 1"
    # message = "PM 56" 
    # message = "SP -0.5"
    # message = "CLDMA"
    # message = "ASC 350000000, 360000000, 10;"
    # message = "AM .55"
    # message = "CLLOG"
    # message = "RDLOG"
    # message = "END"
    # message = "Q?"
    # message = "SAVE"
    # message = "SHUTDOWN"
    message = "HELP"

    client.sendall(f"{message};".encode("ascii"))
    response = client.recv(4096).decode("ascii")
    print(f"[RESPONSE] {response}")

except Exception as e:
    print(f"[ERROR] {e}")
finally:
    client.close()
    print("[INFO] Client disconnected")
```
